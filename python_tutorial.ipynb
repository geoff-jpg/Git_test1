# Python Tutorial

A hands-on introduction to Python covering the core concepts.

## 1. Variables and Data Types

```python
# Integers and floats
age = 30
height = 5.9

# Strings
name = "Alice"

# Booleans
is_student = True

print(type(age), type(height), type(name), type(is_student))
```

```python
# String formatting
print(f"Name: {name}, Age: {age}, Height: {height}")

# String methods
print(name.upper())
print(name.replace("A", "B"))
print(len(name))
```

## 2. Collections: Lists, Tuples, Dicts, Sets

```python
# List — ordered, mutable
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print(fruits)
print(fruits[1])   # indexing
print(fruits[-1])  # last item
print(fruits[1:3]) # slicing
```

```python
# Tuple — ordered, immutable
coordinates = (10, 20)
x, y = coordinates  # unpacking
print(x, y)
```

```python
# Dictionary — key/value pairs
person = {"name": "Alice", "age": 30, "city": "London"}
print(person["name"])
person["email"] = "alice@example.com"  # add key
print(person.get("phone", "not found"))  # safe lookup
print(person.keys())
```

```python
# Set — unordered, unique values
colors = {"red", "green", "blue", "red"}
print(colors)  # duplicates removed
colors.add("yellow")
print("red" in colors)
```

## 3. Control Flow

```python
# if / elif / else
score = 75

if score >= 90:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 50:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
```

```python
# for loop
for fruit in fruits:
    print(fruit)

# range
for i in range(5):
    print(i, end=" ")
```

```python
# while loop
count = 0
while count < 5:
    print(count, end=" ")
    count += 1
```

```python
# List comprehension — compact way to build lists
squares = [x**2 for x in range(1, 6)]
print(squares)

evens = [x for x in range(20) if x % 2 == 0]
print(evens)
```

## 4. Functions

```python
def greet(name, greeting="Hello"):
    """Return a greeting string."""
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", greeting="Hi"))
```

```python
# *args and **kwargs
def summarise(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

summarise(1, 2, 3, name="Alice", city="London")
```

```python
# Lambda functions — anonymous one-liners
double = lambda x: x * 2
print(double(5))

numbers = [3, 1, 4, 1, 5, 9, 2]
print(sorted(numbers, key=lambda x: -x))  # sort descending
```

## 5. Classes and Objects

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}"


class Dog(Animal):  # inheritance
    def fetch(self, item):
        return f"{self.name} fetches the {item}!"


dog = Dog("Rex", "Woof")
print(dog.speak())
print(dog.fetch("ball"))
```

## 6. Error Handling

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError as e:
        return f"Type error: {e}"
    finally:
        print("Division attempted")  # always runs

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "a"))
```

## 7. Working with Files

```python
# Write a file
with open("example.txt", "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")
    f.write("Line 3\n")

# Read it back
with open("example.txt", "r") as f:
    for line in f:
        print(line.strip())
```

## 8. Useful Standard Library Modules

```python
import math
import random
from datetime import datetime

print(math.pi)
print(math.sqrt(144))

print(random.randint(1, 10))
print(random.choice(fruits))

now = datetime.now()
print(now.strftime("%Y-%m-%d %H:%M"))
```
